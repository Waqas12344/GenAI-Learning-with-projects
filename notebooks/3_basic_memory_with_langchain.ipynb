### Manage History

```python
history = []
while True:
    query = input("User: ")
    if query.lower() in ["exit", "quit", "q", "bye"]:
        break
    history.append({"role": "user", "content": query})

    print("User ", query)

    res = llm.invoke(history)
    history.append({"role": "assistant", "content": res.content})
    print("Assistant: ", res.content,"\n")
    print("History: ", history,"\n")
```
