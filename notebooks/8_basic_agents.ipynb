```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int):
    """
    It Will return the sum of two numbers.

    Args:
        a : Number One
        b : Number Two
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int):
    """
    It Will return the product of two numbers.

    Args:
        a : Number One
        b : Number Two
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 3})
multiply_numbers.invoke({"a": 5, "b": 3})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-3.1-flash-lite")
```

```python
agent = create_agent(
    model= llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
response = agent.invoke({"messages":[{"role":"user", "content":"What is the sum of 5 and 13 * 10?"}]})
```

```python
for res in response["messages"]:
    print(res)
    print("\n")
```
