```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import GoogleGenerativeAI
llm = GoogleGenerativeAI(model="gemini-3.1-flash-lite", streaming = True)
```

```python
res = llm.stream("can you please tell me GenAI  in detail")
for chunk in res:
    print(chunk, end="")
```

```python
from langchain_groq import ChatGroq
```

```python
llmgroq = ChatGroq(model="openai/gpt-oss-120b", streaming = True)
```

```python
question = "Can you explain me GenAI"
res = llmgroq.stream(question)
for chunk in res:
    print(chunk.content, end="")
```
